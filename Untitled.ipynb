```python
# Outlier Detection
plt.figure(figsize=(18,12))

df[num_cols].boxplot(rot=90)

plt.title("Outlier Detection")

plt.show()

# Scale problem (Most important) The variables are on very different scales, for example:
# MonthlyIncome: ~1,000 to 20,000
# MonthlyRate: ~2,000 to 27,000
# Age: 18–60
# JobLevel: 1–5
# Attrition: 0–1
# Because MonthlyIncome and MonthlyRate have much larger values, all the other variables appear compressed near zero

# MonthlyIncome Has many outliers (black circles). Most employees earn between approximately 3,000–8,000.
# Some employees earn 17,000–20,000, which are detected as outliers.
# This is normal in HR datasets because senior employees earn much more

# MonthlyRate Very wide spread. Median around 14,000. Maximum around 27,000.
# No major issue; it simply has a larger numerical range

# DailyRate Spread is between roughly 100–1500.
# Doesn't show many extreme outliers

# Other variables look like a straight line because their values are tiny compared to MonthlyIncome and MonthlyRate

# For the IBM HR Employee Attrition dataset (or similar HR datasets), do not remove MonthlyIncome outliers because they 
# represent genuine high-salary employees (managers, directors, etc.)
```

```python
# Remove Extreme Outliers (IQR)
df_clean = df.copy()
```

```python
for col in num_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    df_clean = df_clean[
        (df_clean[col] >= lower) &
        (df_clean[col] <= upper)
    ]

# Note: If too many rows are removed, use winsorization (capping) instead of deleting outliers
```

```python
print("Original Shape :", df.shape)
```

```python
print("New Shape :", df_clean.shape)
# Rows removed: 877 (about 60% of the data)
```

```python
# Winsorization (Capping) (Instead of deleting rows, cap extreme values)

from scipy.stats.mstats import winsorize

df_cap = df.copy()

continuous_cols = [
    "Age",
    "DailyRate",
    "DistanceFromHome",
    "HourlyRate",
    "MonthlyIncome",
    "MonthlyRate",
    "NumCompaniesWorked",
    "PercentSalaryHike",
    "TotalWorkingYears",
    "TrainingTimesLastYear",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsSinceLastPromotion",
    "YearsWithCurrManager"
]

for col in continuous_cols:
    df_cap[col] = winsorize(df_cap[col], limits=[0.01, 0.01])

# This caps only the top and bottom 1% of values and keeps all 1470 employees
```

# FEATURE ENGINEERING:

```python
# Promotion Gap Ratio
df["PromotionGapRatio"] = np.where(
    df["YearsAtCompany"] == 0,
    0,
    df["YearsSinceLastPromotion"] /
    df["YearsAtCompany"]
)

# If YearsAtCompany is 0, then: would cause a division by zero error.
# So this line: np.where(df["YearsAtCompany"] == 0, 0, ...) sets the ratio to 0 whenever YearsAtCompany is 0
```

```python
print(df[["YearsAtCompany",
          "YearsSinceLastPromotion",
          "PromotionGapRatio"]].head(10))

# | YearsAtCompany | YearsSinceLastPromotion | PromotionGapRatio | Meaning                              |
# | -------------: | ----------------------: | ----------------: | ------------------------------------ |
# |             10 |                       2 |              0.20 | Promoted fairly recently.            |
# |             10 |                       8 |              0.80 | Long time since last promotion.      |
# |              5 |                       5 |              1.00 | No promotion during the entire stay. |
# |              1 |                       0 |              0.00 | Recently joined or just promoted.    |
# |              0 |                       0 |              0.00 | Avoids division by zero.             |
```

```python
# Distribution
plt.figure(figsize=(7,4))

sns.histplot(
    df["PromotionGapRatio"],
    bins=20,
    kde=True
)

plt.title("Promotion Gap Ratio")
plt.show()
```

```python
# Role Stagnation Index
df["RoleStagnationIndex"] = np.where(
    df["YearsAtCompany"] == 0,
    0,
    df["YearsInCurrentRole"] /
    df["YearsAtCompany"]
)
```

```python
print(df[["YearsAtCompany",
          "YearsInCurrentRole",
          "YearsAtCompany"]].head(10))
```

```python
# Distribution
plt.figure(figsize=(7,4))

sns.histplot(
    df["RoleStagnationIndex"],
    bins=20,
    kde=True
)

plt.title("Role Stagnation Index")
plt.show()
```

```python
# Training Intensity Score
df["TrainingIntensityScore"] = np.where(
    df["YearsAtCompany"] == 0,
    0,
    df["TrainingTimesLastYear"] /
    df["YearsAtCompany"]
)
```

```python
df[["TrainingTimesLastYear",
    "YearsAtCompany",
    "TrainingIntensityScore"]].head(10)
```

```python
# Distribution
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(7,4))
sns.histplot(df["TrainingIntensityScore"], bins=20, kde=True)
plt.title("Training Intensity Score")
plt.show()
```

```python
# Manager Stability Indicator
df["ManagerStabilityIndicator"] = np.where(
    df["YearsAtCompany"] == 0,
    0,
    df["YearsWithCurrManager"] /
    df["YearsAtCompany"]
)
```

```python
df[["YearsWithCurrManager",
    "YearsAtCompany",
    "ManagerStabilityIndicator"]].head(10)
```

```python
# Distribution
plt.figure(figsize=(7,4))
sns.histplot(df["ManagerStabilityIndicator"], bins=20, kde=True)
plt.title("Manager Stability Indicator")
plt.show()
```

```python
# Career Growth Score
df["CareerGrowthScore"] = (
    df["JobLevel"]*2
    +
    df["PercentSalaryHike"]/10
    +
    df["PerformanceRating"]
    +
    df["TrainingTimesLastYear"]
    -
    df["YearsSinceLastPromotion"]
)
```

```python
df[["JobLevel",
    "PercentSalaryHike",
    "PerformanceRating",
    "TrainingTimesLastYear",
    "YearsSinceLastPromotion",
    "CareerGrowthScore"]].head(10)
```

```python
# Distribution
plt.figure(figsize=(7,4))
sns.histplot(df["CareerGrowthScore"], bins=20, kde=True)
plt.title("Career Growth Score")
plt.show()
```

```python
# Promotion Delay Category
conditions = [

    df["YearsSinceLastPromotion"] <= 1,

    (df["YearsSinceLastPromotion"] > 1) &
    (df["YearsSinceLastPromotion"] <=3),

    (df["YearsSinceLastPromotion"] >3) &
    (df["YearsSinceLastPromotion"] <=6),

    df["YearsSinceLastPromotion"] >6

]

labels = [

    "Recent",

    "Moderate",

    "Delayed",

    "Critical"

]

df["PromotionDelayCategory"] = np.select(
    conditions,
    labels,
    default="Unknown"
)
```

```python
# Plot
plt.figure(figsize=(6,4))
sns.countplot(
    x="PromotionDelayCategory",
    data=df,
    order=["Recent","Moderate","Delayed","Critical"]
)
plt.title("Promotion Delay Category")
plt.show()
```

```python
# Training Need Indicator
df["TrainingNeedIndicator"] = np.where(
    df["TrainingTimesLastYear"] < 2,
    1,
    0
)
```

```python
# Plot
plt.figure(figsize=(5,4))
sns.countplot(x="TrainingNeedIndicator", data=df)
plt.title("Training Need Indicator")
plt.show()
```

```python
# Career Stage
conditions = [

    df["YearsAtCompany"] <3,

    (df["YearsAtCompany"]>=3) &
    (df["YearsAtCompany"]<7),

    (df["YearsAtCompany"]>=7) &
    (df["YearsAtCompany"]<15),

    df["YearsAtCompany"]>=15

]

labels = [

    "Early",

    "Mid",

    "Senior",

    "Veteran"

]

df["CareerStage"] = np.select(
    conditions,
    labels,
    default="Unknown"
)
```

```python
# Plot
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(7,4))
sns.countplot(
    x="CareerStage",
    data=df,
    order=["Early", "Mid", "Senior", "Veteran"]
)

plt.title("Career Stage Distribution")
plt.show()
```

```python
# High Promotion Gap Flag
df["HighPromotionGap"] = np.where(
    df["YearsSinceLastPromotion"] >=5,
    1,
    0
)
```

```python
# Plot
plt.figure(figsize=(5,4))
sns.countplot(x="HighPromotionGap", data=df)

plt.title("High Promotion Gap Flag")
plt.show()
```

# CAREER PROGRESSION PATTERNS :
# (using unsupervised learning (K-Means (primary) and Hierarchical Clustering (validation))

```python
# inport libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from scipy.cluster.hierarchy import dendrogram, linkage, fcluster

from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings("ignore")

plt.style.use("ggplot")
pd.set_option("display.max_columns", None)
```

```python
# Load Dataset
df = pd.read_csv("Processed_Data.csv")

df.head()
```

```python
print(df.shape)
```

```python
# Select Features for Clustering (Only career-related variables should be used)

career_features = [

    "YearsAtCompany",

    "YearsInCurrentRole",

    "YearsSinceLastPromotion",

    "YearsWithCurrManager",

    "PromotionGapRatio",

    "RoleStagnationIndex",

    "TrainingIntensityScore",

    "ManagerStabilityIndicator",

    "CareerGrowthScore",

    "TrainingTimesLastYear",

    "JobLevel"

]

X = df[career_features]

X.head()
```

```python
# Feature Scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
print(X_scaled.shape)
```

```python
# Elbow Method
inertia = [] 
# ek empty list banata hai. Is list me har value of K (number of clusters) ke liye Inertia (WCSS) store hogi. Inertia batati hai ki data points apne 
# cluster center (centroid) ke kitne close hain. Lower inertia = Better clustering.

K = range(2,11)
# matlab hai: 2, 3, 4, 5, 6, 7, 8, 9, 10 Hum 2 se 10 clusters tak try karenge. Har K ke liye K-Means model train hoga

for k in K:

    model = KMeans(
        n_clusters=k,  # Kitne clusters banane hain. Agar k = 4 hai to data ko 4 groups me divide karega
        random_state=42,  # K-Means randomly centroids choose karta hai. random_state=42 dene se har baar same result milega
        n_init=10  # K-Means ek baar nahi chalti. Yeh 10 different random centroid positions se start karti hai.
                   # Phir sabme se jiski inertia sabse kam hoti hai us solution ko choose karti hai
    )

    model.fit(X_scaled)

    inertia.append(model.inertia_)  # Training ke baad model ki inertia nikalta hai. Har inertia value list me add ho jati hai

# eg:
# | K | Inertia |
# | - | ------- |
# | 2 | 9500    |
# | 3 | 7000    |
# | 4 | 5200    |
# | 5 | 4300    |
```

```python
plt.figure(figsize=(8,5))

plt.plot(K, inertia, marker="o")

plt.xlabel("Number of Clusters")

plt.ylabel("WCSS")

plt.title("Elbow Method")

plt.show()
```

```python
# Silhouette Score
# Silhouette Score batata hai: Clusters kitne achhe bane hain

scores = []

for k in range(2,11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    scores.append(score)

    print(k, score)

# | Score          | Meaning              |
# | -------------- | -------------------- |
# | **1**          | Excellent clustering |
# | **0.7 - 0.9**  | Very good            |
# | **0.5 - 0.7**  | Good                 |
# | **0.25 - 0.5** | Average              |
# | **<0.25**      | Poor clustering      |
# | **0**          | Overlapping clusters |
# | **Negative**   | Wrong clustering     |

# Higher score = Better clustering
```

```python
# Best Number of Clusters
best_k = range(2,11)[np.argmax(scores)]

print("Best K :", best_k)
```

```python
# Train Final KMeans Model
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["CareerCluster"] = kmeans.fit_predict(
    X_scaled
)
```

```python
# Cluster Distribution
df["CareerCluster"].value_counts().sort_index()
```

```python
# plt.figure(figsize=(7,5))

sns.countplot(
    x="CareerCluster",
    data=df
)

plt.title("Career Cluster Distribution")

plt.show()
```

```python
# PCA Visualization
pca = PCA(n_components=2)

components = pca.fit_transform(X_scaled)

df["PC1"] = components[:,0]

df["PC2"] = components[:,1]
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(

    data=df,

    x="PC1",

    y="PC2",

    hue="CareerCluster",

    palette="Set2",

    s=70

)

plt.title("Career Clusters (PCA)")

plt.show()

# This indicates that K-Means has successfully separated employees into two distinct career groups.
# The limited overlap suggests that employees within each cluster are relatively similar to one another and different from employees in the other cluster
```

```python
# Hierarchical Clustering
linked = linkage(
    X_scaled,
    method="ward"
)
```

```python
plt.figure(figsize=(14,6))

dendrogram(   # A dendrogram shows how employees or groups of employees are merged based on similarity

    linked,

    truncate_mode="lastp",  # Sirf last merged clusters dikhaye jayenge. Isse graph readable ban jata hai

    p=20,  # Matlab: Last 20 clusters/leaves show karo

    leaf_rotation=90  # Leaves (X-axis labels) ko 90° rotate karta hai. Without rotation: Employee1 Employee2 Employee3 Employee4
                      # Sab overlap karenge. After rotation:
                      # Employee1
                      # Employee2
                      # Employee3
                      # Employee4  Read karna easy ho jata hai.

)

plt.title("Hierarchical Clustering Dendrogram")

plt.show()

# X-axis = Each label represents an employee (or, depending on the plotting settings, a sampled observation).
# Y-axis = The height represents the distance (dissimilarity) between clusters. Lower merge height = more similar. Higher merge height = less similar.
# Observation = there are two major branches: Left branch Right branch. These merge only at a relatively high distance (around 82).
# Interpretation:
# This indicates that the data naturally forms two major groups, which supports your K-Means solution with K = 2.
# Why is the last merge high? The final merge occurs at the highest distance because the two groups are quite different from each other.
# This is exactly what you want in clustering.
```

```python
# Hierarchical Cluster Labels
df["HC_Cluster"] = fcluster(

    linked,

    best_k,

    criterion="maxclust"  # criterion batata hai ki clustering kis rule se karni hai. Yahan: criterion="maxclust" ka matlab hai: Maximum number of clusters = best_k

)
```

# PROMOTION GAP RISK SCORING:

```python
# Import Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
```

```python
# Load Clustered Dataset
df = pd.read_csv("Clustered_Career_Data.csv")

df.head()
```

```python
print(df.shape)
```

```python
# Normalize Career Features
from sklearn.preprocessing import MinMaxScaler

career_cols = [

    "PromotionGapRatio",

    "RoleStagnationIndex",

    "TrainingIntensityScore",

    "ManagerStabilityIndicator"

]

scaler = MinMaxScaler()

df[career_cols] = scaler.fit_transform(df[career_cols])
```

```python
# Promotion Gap Score
# Higher score = Higher career stagnation risk

df["PromotionGapScore"] = (

      0.40 * df["PromotionGapRatio"]

    + 0.30 * df["RoleStagnationIndex"]

    + 0.20 * (1 - df["TrainingIntensityScore"])

    + 0.10 * (1 - df["ManagerStabilityIndicator"])

)
```

```python
# Promotion Gap Risk Level
conditions = [
    df["PromotionGapRatio"] < 0.25, # Check karta hai: Kya Promotion Gap Ratio 0.25 se kam hai? Agar Yes, to employee Low Risk category me jayega

    (df["PromotionGapRatio"] >= 0.25) &
    (df["PromotionGapRatio"] < 0.50), # Yeh check karta hai: Ratio 0.25 ya usse zyada ho Aur 0.50 se kam ho Dono conditions true honi chahiye

    (df["PromotionGapRatio"] >= 0.50) &
    (df["PromotionGapRatio"] < 0.75),

    df["PromotionGapRatio"] >= 0.75
]

labels = [  # Har condition ka ek corresponding label hai
    "Low",
    "Medium",
    "High",
    "Critical"
]

df["PromotionGapRisk"] = np.select(  # np.select() har employee ke liye conditions ko upar se neeche check karta hai.
    conditions,
    labels,
    default="Unknown"
)
df["PromotionGapRisk"]
```

```python
# Career Growth Level
conditions = [

    df["CareerGrowthScore"] >=15,

    (df["CareerGrowthScore"]>=10) &
    (df["CareerGrowthScore"]<15),

    (df["CareerGrowthScore"]>=6) &
    (df["CareerGrowthScore"]<10),

    df["CareerGrowthScore"]<6

]

labels = [

    "Excellent",

    "Good",

    "Average",

    "Poor"

]

df["CareerGrowthLevel"] = np.select(
    conditions,
    labels,
     default="Unknown"
)
df["CareerGrowthLevel"]
```

```python
# Training Need Level
conditions = [

    df["TrainingTimesLastYear"] >=5,

    (df["TrainingTimesLastYear"]>=3) &
    (df["TrainingTimesLastYear"]<5),

    (df["TrainingTimesLastYear"]>=1) &
    (df["TrainingTimesLastYear"]<3),

    df["TrainingTimesLastYear"]==0

]

labels = [

    "Low Need",

    "Moderate Need",

    "High Need",

    "Urgent Need"

]

df["TrainingNeedLevel"] = np.select(
    conditions,
    labels,
    default="Unknown"
)
df["TrainingNeedLevel"]
```

```python
# Manager Stability Impact
conditions = [

    df["YearsWithCurrManager"]>=8,

    (df["YearsWithCurrManager"]>=5) &
    (df["YearsWithCurrManager"]<8),

    (df["YearsWithCurrManager"]>=2) &
    (df["YearsWithCurrManager"]<5),

    df["YearsWithCurrManager"]<2

]

labels = [

    "Very Stable",

    "Stable",

    "Moderate",

    "Unstable"

]

df["ManagerStabilityImpact"] = np.select(
    conditions,
    labels,
    default="Unknown"
)
df["ManagerStabilityImpact"]
```

```python
# Retention Opportunity Index
# Higher = Higher intervention priority

df["RetentionOpportunityIndex"] = (

      40 * df["PromotionGapScore"]

    + 20 * (1-df["TrainingIntensityScore"])

    + 20 * df["RoleStagnationIndex"]

    + 20 * (1-df["ManagerStabilityIndicator"])

)
df["RetentionOpportunityIndex"]
```

```python
# Retention Priority
conditions = [

    df["RetentionOpportunityIndex"]<25,

    (df["RetentionOpportunityIndex"]>=25) &
    (df["RetentionOpportunityIndex"]<50),

    (df["RetentionOpportunityIndex"]>=50) &
    (df["RetentionOpportunityIndex"]<75),

    df["RetentionOpportunityIndex"]>=75

]

labels = [

    "Low",

    "Medium",

    "High",

    "Critical"

]

df["RetentionPriority"] = np.select(
    conditions,
    labels,
    default="Unknown"
)
df["RetentionPriority"]
```

```python
# Suggested HR Action
def suggest_action(row):

    if row["RetentionPriority"]=="Critical":
        return "Immediate Promotion Review"

    elif row["PromotionGapRisk"]=="High":
        return "Career Development Plan"

    elif row["TrainingNeedLevel"]=="Urgent Need":
        return "Mandatory Skill Training"

    elif row["ManagerStabilityImpact"]=="Unstable":
        return "Manager Mentorship Review"

    else:
        return "Regular Career Monitoring"

df["SuggestedAction"] = df.apply(
    suggest_action,
    axis=1
)
df["SuggestedAction"]
```
